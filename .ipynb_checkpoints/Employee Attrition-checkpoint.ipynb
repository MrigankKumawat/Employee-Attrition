### Machine Learning

```python
x = df.drop(['Attrition', 'Attrition_num'], axis=1)
y = df['Attrition']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
)
```

### Separating Features and Lables

```python
attrition_features_train = x_train.copy()
attrition_labels_train = y_train.copy()
attrition_labels_train = attrition_labels_train.map({'No': 0, 'Yes': 1})

attrition_features_test = x_test.copy()
attrition_labels_test = y_test.copy()
attrition_labels_test = attrition_labels_test.map({'No': 0, 'Yes': 1})
```

```python
### Separating Numerical and Categorical Features
```

```python
attrition_num_train = attrition_features_train.select_dtypes(include=[np.number]).columns.tolist()
attrition_cat_train = attrition_features_train.select_dtypes(include=['object', 'bool', 'category']).columns.tolist()
# attrition_num_train = att.select_dtypes(include=[np.number])
# attrition_cat_train = attrition_features_train.select_dtypes(include=['object', 'bool', 'category'])
```

### Pipelines

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
cat_pipeline = Pipeline([
    ("onehot", OneHotEncoder(handle_unknown="ignore")),
])
full_pipeline = ColumnTransformer([
    ("num", num_pipeline, attrition_num_train),
    ("cat", cat_pipeline, attrition_cat_train)
])
```

### Transforming the Pipeline

```python
attrition_prepared_train = full_pipeline.fit_transform(attrition_features_train)
```

## Model Training

### Random Forest Classifier

```python
rand_forest_model = RandomForestClassifier()
rand_forest_model.fit(attrition_prepared_train, attrition_labels_train)
```

```python
forest_predict_train = rand_forest_model.predict(attrition_prepared_train)
```

```python
forest_accuracy = accuracy_score(attrition_labels_train, forest_predict_train)
forest_prec = precision_score(attrition_labels_train, forest_predict_train)
forest_recall = recall_score(attrition_labels_train, forest_predict_train)
forest_f1 = f1_score(forest_predict_train, attrition_labels_train)

print(f"Accuracy Score:{forest_accuracy:.2f}")
print(F"Precision Score:{forest_prec:.2f}")
print(f"Recall Score:{forest_recall:.2f}")
print(f"F1 Score:{forest_f1:.2f}")
# print(f"Classification Report:{classification_report(depress_train_label, forest_predict_train)}")
```

### Testing Set 

```python
attrition_prepared_test = full_pipeline.transform(attrition_features_test)
```

```python
forest_predict_test = rand_forest_model.predict(attrition_prepared_test)

accuracy_score(attrition_labels_test, forest_predict_test)
precision_score(attrition_labels_test, forest_predict_test)
recall_score(attrition_labels_test, forest_predict_test)
f1_score(attrition_labels_test, forest_predict_test)
```
